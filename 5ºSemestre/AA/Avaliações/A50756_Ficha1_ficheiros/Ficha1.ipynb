## Imports

```python
import matplotlib.pyplot as plt
import numpy as np
import pickle
import scipy.spatial.distance as spd
```

```python
with open("A50756_Q001_data.p", "rb") as data_file:
    D = pickle.load(data_file)


with open("A50756_Ficha1_Respostas.p", "rb") as respostas_file:
    respostas = pickle.load(respostas_file)
```

```python
print(D)
print(D.keys())
print(respostas)
```

## 1.
### a)

F, F, T, F

```python
X = D['dados']
y = D['trueClass']


print("X shape:", X.shape)
print("y shape:", y.shape)

print("Categorias:", np.unique(y))
```

```python
X0 = X[:, y==0]

m0 = np.mean(X0, axis=1)

print("i:", np.round(m0, 2))

X1 = X[:, y==1]

m1 = np.mean(X1, axis=1)

print("ii:", np.round(m1, 2))
print()

respostas['Q001'][0,:] = [0,0,1,0] 
print(respostas['Q001'][0,:])
```

### b)

F, T, F, F

```python
X2 = X[:, y==2]
X3 = X[:, y==3]

m2 = np.mean(X2, axis=1)
m3 = np.mean(X3, axis=1)

manhattan = np.sum(np.abs(m2-m3))

print("i:", np.round(manhattan, 2))

cossine = spd.cosine(m0, m1)

print("ii:", np.round(cossine, 2))


respostas['Q001'][1,:] = [0,1,0,0] 
print(respostas['Q001'][1,:])
```

### c)

F, F, T, F

```python
P0 = np.sum(y == 0) / len(y)

print("i:", np.round(P0, 2))
print()

covX2 = np.cov(X2)
print("ii:", np.round(covX2, 2))
print()


respostas['Q001'][2,:] = [0,0,1,0] 
print(respostas['Q001'][2,:])
```

### d)
T, F, F, F

```python
media = np.mean(X, axis=1)
print("i:", np.round(media, 2))
print()

cov = np.cov(X)
print("ii:")
print(np.round(cov, 2))

respostas['Q001'][3,:] = [1,0,0,0] 
print()

print(respostas['Q001'][3,:])
```

## 2
### a)

F, T, F, F

```python
X = np.array([
    [-4, -5, -4, 5, 3, 3],
    [-1, -2, 1, -1, -1, 0]
])

X0 = X[:, :3]
X1 = X[:, 3:]

m0 = np.mean(X0, axis=1)
m1 = np.mean(X1, axis=1)

produto = np.dot(m0, m1)
print("i:", np.round(produto, 2))

norma0 = np.linalg.norm(m0)
print("ii:", np.round(norma0, 2))

respostas['Q002'][0,:] = [0,1,0,0] 
print(respostas['Q002'][0,:])
```

### b)

F, T, F, F

```python
produto = np.dot(X0, X0.T)
print("i:", np.round(produto, 2))
print()

cov0 = np.cov(X0)
print("ii:", np.round(cov0, 2))
print()

respostas['Q002'][1,:] = [0,1,0,0] 
print(respostas['Q002'][1,:])
```
