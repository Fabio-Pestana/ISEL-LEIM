## b)

### F, F, T, F

```python
X1 = X[:, y==1]
X2 = X[:, y==2]


covX1 = np.cov(X1)
covX2 = np.cov(X2)

prodX12 = np.dot(covX1, covX2)

det_i = np.linalg.det(prodX12)

print(np.round(det_i, 2))
```

```python
X1 = X[:, y==1]
X2 = X[:, y==2]

m1 = np.mean(X1, axis=1)
m2 = np.mean(X2, axis=1)

covX2 = np.cov(X2)

prod_aux = np.dot(m1.T, covX2)
prod_final = np.dot(prod_aux, m2)

print(np.round(prod_final, 2))
```

```python
X1 = X[:, y==1]
X3 = X[:, y==3]

m3 = np.mean(X3, axis=1)

covX1 = np.cov(X1)

prod_final = np.dot(covX1, m3)

print(np.round(prod_final, 2))
```

```python
X1 = X[:, y==1]
X3 = X[:, y==3]

m1 = np.mean(X1, axis=1)
m3 = np.mean(X3, axis=1)

prod_final = np.dot(m1, m3)

print(np.round(prod_final, 2))
```

## c)

### F, F, F, T

```python
X0 = X[:, y==0]
X1 = X[:, y==1]

m0 = np.mean(X0, axis=1)
m1 = np.mean(X1, axis=1)

distancia_Manhattan = np.sum(np.abs(m0 - m1))

print(round(distancia_Manhattan,2))
```

```python
X0 = X[:, y==0]
X1 = X[:, y==1]

m0 = np.mean(X0, axis=1)
m1 = np.mean(X1, axis=1)

prod = np.dot(m0, m1)
M0n = np.linalg.norm(m0)
M1n = np.linalg.norm(m1)
prod_normais = M0n * M1n

distancia_cosseno = 1 - (prod/prod_normais)

print(round(distancia_cosseno,2))
```
