# 1 a)

```python
X = D['X']
y = D['y']

print("X shape:", X.shape)  # (2, N)
print("y shape:", y.shape)  # (N,)

# Vetor de pesos fornecido
w = np.array([0.00, -0.15, -0.99])  # w0, w1, w2

# Transpor X para que tenha a forma correta
X = X.T  # De (2, N) para (N, 2)

# Adicionar x0 = 1 para todas as amostras (termo bias)
X_with_bias = np.hstack((np.ones((X.shape[0], 1)), X))  # Forma final: (N, 3)

# Verificando as formas antes da multiplicação
print("Forma de X_with_bias:", X_with_bias.shape)  # Deve ser (N, 3)
print("Forma de w:", w.shape)  # Deve ser (3,)

# Calcular as predições lineares
y_hat = np.dot(X_with_bias, w)  # Produto escalar para obter predições lineares

# Converter predições em classes (+1 para y_hat >= 0, -1 para y_hat < 0)
y_pred = np.where(y_hat >= 0, 1, -1)

# Adaptar y_pred para o formato exigido (0 e 1)
y_pred = np.where(y_pred == -1, 0, 1)  # Converte -1 para 0, mantém 1 como 1

# Calcular o erro absoluto médio
erro_absoluto_medio = np.mean(np.abs(y - y_pred))

# Calcular o número de acertos na classe ω1 (positivos)
acertos_w1 = np.sum((y_pred == 1) & (y == 1))

# Exibir os resultados
print("Número de acertos na classe ω1:", acertos_w1)
print("Erro absoluto médio:", np.round(erro_absoluto_medio, 3))

# iv - V

respostas['Q001'][0,:] = [0,0,0,1] 
respostas['Q001'][0,:]
```

```python
precision = precision_score(y, y_pred, average='binary')
recall = recall_score(y, y_pred, average='binary')
cm = confusion_matrix(y, y_pred)

# Extrair valores da matriz de confusão
TN, FP, FN, TP = cm.ravel()

# Número de acertos na classe positiva (TP)
num_correct_positive = TP

# Taxa de falsos alarmes
false_alarm_rate = FP / (FP + TN)

print(np.round(precision,3))
print(np.round(num_correct_positive,3))
print(np.round(recall,3))
print(np.round(false_alarm_rate, 3))
print(cm)


respostas['Q001'][1,:] = [1,0,0,0] 
respostas['Q001'][1,:]
```

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import Lasso
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error
# Separar as variáveis
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import PolynomialFeatures

# Carregar o conjunto de dados
data = load_diabetes()
X, y = data.data, data.target

# Dividir os dados em treino (primeiras 312 amostras) e teste (restantes)
X_train, y_train = X[:165], y[:165]
X_test, y_test = X[165:], y[165:]

# Gerar termos polinomiais de 2ª ordem
poly = PolynomialFeatures(degree=2)

X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)

# Ajustar o modelo de regressão linear
model = LinearRegression()
model.fit(X_train_poly, y_train)

# Previsões no conjunto de teste
y_test_pred = model.predict(X_test_poly)
y_train_pred = model.predict(X_train_poly)


# Métricas
mse_test = mean_squared_error(y_test, y_test_pred)
erro_absoluto_medio = np.mean(np.abs(y_train - y_train_pred))

print(mse_test)
print(erro_absoluto_medio)


respostas['Q002'][0,:] = [0,0,0,1] 
respostas['Q002'][0,:]
```

## b)

```python
# Gerar termos polinomiais de 3ª ordem
poly = PolynomialFeatures(degree=3)
X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)

# Ajustar o modelo Lasso
lasso = Lasso(random_state=42, alpha=0.01)
lasso.fit(X_train_poly, y_train)

# Previsões no conjunto de teste
y_test_pred = lasso.predict(X_test_poly)
y_treino_pred = lasso.predict(X_train_poly)

# Calcular métricas
erro_treino = np.mean(np.abs(y_train - y_treino_pred))
erro_teste = np.mean(np.abs(y_test - y_test_pred))

print(np.round(erro_treino, 2))
print(np.round(erro_teste, 2))

mae_train = mean_absolute_error(y_train, y_train_pred)
mae_test = mean_absolute_error(y_test, y_test_pred)

# Exibir os resultados
print(f"Erro absoluto médio no conjunto de treino: {mae_train:.2f}")
print(f"Erro absoluto médio no conjunto de teste: {mae_test:.2f}")

respostas['Q002'][1,:] = [0,0,0,1] 
respostas['Q002'][1,:]
```

```python
with open("A50756_Q003_data.p", "rb") as data_file:
    D = pickle.load(data_file)

# Separar as variáveis
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import PolynomialFeatures

x = np.array(D['x']).reshape(-1, 1)
y = np.array(D['y'])
folds = np.array(D['folds'])

# Criar os conjuntos de treino e teste
train_indices = folds == 0
test_indices = folds == 1

x_train, y_train = x[train_indices], y[train_indices]
x_test, y_test = x[test_indices], y[test_indices]


poly = PolynomialFeatures(degree=3)
x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.transform(x_test)

model = LinearRegression()
model.fit(x_train_poly, y_train)

# Previsões
y_train_pred = model.predict(x_train_poly)
y_test_pred = model.predict(x_test_poly)

# Métricas
mse_test = mean_squared_error(y_train, y_train_pred)
r2_train = r2_score(y_train, y_train_pred)
coefficients = model.intercept_, *model.coef_

print(np.round(mse_test, 2))
print(np.round(coefficients[0], 2))

respostas['Q003'][0,:] = [0,1,0,0] 
respostas['Q003'][0,:]
```

```python
# Criar os conjuntos de treino e teste
train_indices = folds == 1
test_indices = folds == 0

x_train, y_train = x[train_indices], y[train_indices]
x_test, y_test = x[test_indices], y[test_indices]

# Regressão polinomial de 4ª ordem
poly = PolynomialFeatures(degree=4)
x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.transform(x_test)

model = LinearRegression()
model.fit(x_train_poly, y_train)

# Previsões
y_train_pred = model.predict(x_train_poly)
y_test_pred = model.predict(x_test_poly)

# Métricas
r2_train = r2_score(y_train, y_train_pred)
r2_test = r2_score(y_test, y_test_pred)

print(np.round(r2_train, 2))
print(np.round(r2_test, 2))

respostas['Q003'][1,:] = [0,0,0,1] 
respostas['Q003'][1,:]
```
