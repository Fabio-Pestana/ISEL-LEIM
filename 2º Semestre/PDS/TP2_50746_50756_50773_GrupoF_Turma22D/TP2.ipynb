```python
from scipy.fft import fft

def calcular_caract(array_sinais, taxa_amostragem):
    parametros_sinais = []

    for sinal in array_sinais:
        # Parâmetros do domínio do tempo
        amplitude_media = np.mean(np.abs(sinal))
        desvio_padrao_tempo = np.std(sinal)
        valor_max_tempo = np.max(sinal)
        valor_min_tempo = np.min(sinal)

        # Transformada de Fourier
        transformada = fft(sinal)
        espectro = np.abs(transformada)
        frequencias = np.fft.fftfreq(len(sinal), 1 / taxa_amostragem)

        # Parâmetros do domínio da frequência
        valor_pico = np.max(espectro)
        indice_pico = np.argmax(espectro)

        if indice_pico < len(frequencias):
            pico_frequencia = np.abs(frequencias[indice_pico])
        else:
            pico_frequencia = 0

        # Adicionar os parâmetros calculados à lista de resultados
        parametros_sinais.append({
            "amplitude_media": np.round(amplitude_media,5),
            "desvio_padrao_tempo": np.round(desvio_padrao_tempo, 5),
            "valor_max_tempo": np.round(valor_max_tempo,5),
            "valor_min_tempo": np.round(valor_min_tempo, 5),
            "pico_frequencia": np.round(pico_frequencia, 5),
            "valor_pico": np.round(valor_pico, 5)
        })

    return parametros_sinais
```

```python
import os
import glob
import numpy as np
from scipy.fft import rfft, rfftfreq
from scipy.io import wavfile
from scipy.signal import find_peaks


def calculate_pitch(audio_file, sample_rate):
    # Carrega o arquivo de áudio
    audio, _ = read_audio(audio_file, sample_rate)
    
    # Calcula a transformada de Fourier (STFT)
    window_size = 2048
    stft = rfft(audio, window_size)
    frequencies = rfftfreq(window_size, d=1/sample_rate)

    # Encontra os picos na magnitude do espectro
    magnitudes = np.abs(stft)
    peaks, _ = find_peaks(magnitudes)

    # Encontra o pico mais alto acima de uma determinada frequência mínima
    min_frequency = 20  # Frequência mínima em Hz (valor mais baixo que pode ser captado pelo ser humano)
    peaks_above_min = [peak for peak in peaks if frequencies[peak] >= min_frequency]
    dominant_peak = peaks_above_min[np.argmax(magnitudes[peaks_above_min])]

    # Calcula o pitch correspondente à frequência dominante
    pitch = frequencies[dominant_peak]

    return pitch

# Função auxiliar para carregar o arquivo de áudio
def read_audio(audio_file, sample_rate):
    len_data = len(audio_file) # comprimento do array do audio
    d = len_data / sample_rate # retorna a duração em float
    duration = d  # Duração em segundos
    t = np.linspace(0, duration, int(duration * sample_rate), endpoint=False)
    audio = 0.3 * np.sin(2 * np.pi * 440 * t)  # Tom de 440 Hz

    return audio, sample_rate


def output_duration(length):
	hours = length // 3600 # horas
	length %= 3600
	mins = length // 60 # minutos
	length %= 60
	seconds = length # segundos

	return hours, mins, seconds

# Exemplo de uso
for file in glob.glob("D:\ISEL\PDS\TP2\*.wav"): #abre a pasta dos audios
    file_name = os.path.basename(file)
    if file_name.split("_")[0] == "Assobio":
        wav = f"D:\ISEL\PDS\TP2\{file_name}"
        samplerate, data = wavfile.read(wav)
        pitch = calculate_pitch(data, samplerate)
        print(file_name, '-> Pitch: {:.2f} Hz'.format(pitch))
```

```python
import scipy.signal as ss
```

```python
def FBCF(g,N):
    a = np.hstack((1,np.zeros(N-1),g))
    b = [1]
    return b,a
```

```python
def AP(g,N):
    a = np.hstack((1,np.zeros(N-1),-g))
    b = np.hstack((-g,np.zeros(N-1),1))
    return b,a
```

## Texto sobre as caracteristicas dos filtros

>Filtro Comb (Comb Filter):

O filtro Comb é usado para criar repetições espaçadas no tempo, simulando o efeito de reflexões sonoras em uma sala. Sua função de transferência é dada por FBCF(g, N) = 1 / (1 - gz^(-N)), onde g é o coeficiente de realimentação e N é o atraso em amostras.

No domínio da frequência, o filtro Comb tem uma resposta em frequência que consiste em uma série de picos espaçados igualmente, chamados de "comb filter notches". A distância entre os picos está relacionada ao atraso do filtro. Cada pico possui uma amplitude reduzida em relação à frequência central.

>Filtro Allpass (Allpass Filter):

O filtro Allpass é usado para alterar a fase do sinal sem afetar sua magnitude. Na reverberação, é usado para simular o espalhamento de frequências causado pelas reflexões. Sua função de transferência é dada por AP(g, N) = (−g + z^(-N)) / (1 - gz^(-N)), onde g é o coeficiente de realimentação e N é o atraso em amostras.

No domínio da frequência, o filtro Allpass tem uma resposta em frequência plana, o que significa que não há alteração na magnitude do sinal. No entanto, o filtro Allpass modifica a fase do sinal. Ele introduz uma resposta de fase que varia com a frequência, causando o espalhamento temporal das frequências.
