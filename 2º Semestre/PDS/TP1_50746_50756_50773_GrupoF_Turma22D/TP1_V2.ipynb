```python

def c(t, N):
    f0 = 5
    result = 0
    for k in range(1, N+1):
        result += np.sin(2*np.pi*(2*k-1)*f0*t)/(2*k-1)
    return 4/np.pi * result

Fs_c= 10500 # Fs > 2*Freq_max neste caso fre_max e 5000
tc = np.arange(0, 1, 1/Fs_c)

# N = 1
c1 = c(tc, 1)

# N = 10
c10 = c(tc, 10)

# N = 1000
c1000 = c(tc, 1000)

# Plot para N = 1
plt.subplot(311)
plt.plot(tc, c1)
plt.title('c(t) para N = 1')
plt.xlabel('t(s)')
plt.ylabel('c(t)')
plt.show()

# Plot para N = 10
plt.subplot(312)
plt.plot(tc, c10)
plt.title('c(t) para N = 10')
plt.xlabel('t(s)')
plt.ylabel('c(t)')
plt.show()

# Plot para N = 1000
plt.subplot(313)
plt.plot(tc, c1000)
plt.title('c(t) para N = 1000')
plt.xlabel('t(s)')
plt.ylabel('c(t)')
plt.show()
```

>**d)**

```python
def d(t, N):
    signal = np.zeros(t.shape, dtype=np.complex128)
    for k in range(-N, N+1):
        signal+=X(k)
    return signal

def X(k):
    f0 = 1
    if k == 0:
        return 1/(3*np.pi**2)
    else:
        return (-1)**k * (2*(1-(-1)**k)) / (k**2 * np.pi**2) * np.exp((0+1j)*2*np.pi*k*f0*t)

Fs_d=2500 #Fs > 2*Freq_max neste caso fre_max é 1000
t = np.arange(0, 2,1/Fs_d)

# N = 1
d1 = d(t, 1)

# N = 10
d10 = d(t, 10)

# N = 1000
d1000 = (d(t, 1000))

# Plot para N = 1
plt.subplot(311)
plt.plot(t, d1)
plt.title('d(t) para N = 1')
plt.xlabel('t(s)')
plt.ylabel('d(t)')
plt.show()

# Plot para N = 10
plt.subplot(312)
plt.plot(t, d10)
plt.title('d(t) para N = 10')
plt.xlabel('t(s)')
plt.ylabel('d(t)')
plt.show()

# Plot para N = 1000
plt.subplot(313)
plt.plot(t, d1000)
plt.title('d(t) para N = 1000')
plt.xlabel('t(s)')
plt.ylabel('d(t)')
plt.show()
```

```python
notas = [('g', 4), ('g', 4), ('g', 4), ('d#', 4), ('g#', 8),
         ('g', 4), ('d#', 4), ('g#', 8), ('g', 2),
         ('c5', 4), ('c5', 4), ('c5', 4), ('d#', 4), ('g#', 8),
         ('g', 4), ('d#', 4), ('g#', 8), ('g', 2)]

bpm = 206

def ADSR(notes, bpm, a=1, d=0.2, s=0.8, r=0.1):
    durationunit = 60 / bpm
    samplerate = 44100
    amplitude = 1
    time = 0
    
    composition = []
    for note, duration in notes:
        frequency = 440 * 2 ** ((ord(note[0]) - ord("a")) / 12)
        if len(note) > 1 and note[1] == "#":
            frequency *= 2 ** (1 / 12)
        elif len(note) > 1 and note[1] == "b":
            frequency /= 2 ** (1 / 12)
        
        # Calcula a duração do ADSR
        attack_dur = duration * a 
        decay_dur = duration * d 
        sustain_dur = duration * s
        release_dur = duration * r
        
        # Cria o envelope
        attack_env = np.linspace(0, amplitude, int(attack_dur * samplerate))
        decay_env = np.exp(-5 * np.linspace(0, 1, int(decay_dur * samplerate)))
        sustain_env = np.ones(int(sustain_dur * samplerate))
        release_env = np.linspace(amplitude, 0, int(release_dur * samplerate))
        envelope = np.concatenate((attack_env, decay_env, sustain_env, release_env))
        num_pad = int(duration * durationunit * samplerate) - len(envelope)
        if num_pad < 0:  # impedir valores negativos
            envelope = envelope[:int(duration * durationunit * samplerate)]
        else:
            envelope = np.pad(envelope, (0, num_pad), 'constant')
        
        # Gera a música com o ADSR
        samples = np.arange(int(duration * durationunit * samplerate))
        waveform =( amplitude * np.sin(2 * np.pi * frequency * samples / samplerate) ) + envelope
        composition.extend(waveform)
    
    # Play audio
    display(Audio(composition, rate=samplerate))
    
    # Visualização do sinal no domínio do tempo
    plt.figure(figsize=(10, 4))
    plt.plot(composition)
    plt.xlim(0,10000)
    plt.title('Composição musical no domínio do tempo')
    plt.xlabel('Tempo (ms)')
    plt.ylabel('Música(t)')
    plt.show()

    # Visualização do espectro do sinal
    plt.figure(figsize=(10, 4))
    plt.magnitude_spectrum(composition, Fs=44100)
    plt.xlim(0,1000)
    plt.xlabel('Frequência (Hz)')
    plt.ylabel('Amplitude')
    plt.title('Espectro da composição musical')
    plt.show()

# Visualização do espectrograma do sinal
    plt.figure(figsize=(10, 4))
    plt.specgram(composition,NFFT=512, Fs=44100)
    plt.xlabel('Tempo (segundos)')
    plt.ylabel('Frequência (Hz)')
    plt.title('Espectrograma da composição musical')
    plt.colorbar()
    plt.show()

composition = ADSR(notas,bpm)
```

```python
import ipywidgets as widgets
import scipy.signal


notas = [('g', 4), ('g', 4), ('g', 4), ('d#', 4), ('g#', 8),
         ('g', 4), ('d#', 4), ('g#', 8), ('g', 2),
         ('c5', 4), ('c5', 4), ('c5', 4), ('d#', 4), ('g#', 8),
         ('g', 4), ('d#', 4), ('g#', 8), ('g', 2)]

bpm = 226

def criar_composicao(notas, bpm, forma_onda):
    # Definir parâmetros do sinal
    fs = 44100  # Frequência de amostragem em Hz
    duracao_unidade = 60 / bpm  # Duração de uma unidade de tempo em segundos
    duracao_total = sum([duracao_unidade * duracao for nota, duracao in notas])  # Duração total em segundos
    amplitude = 1  # Amplitude máxima do sinal

    # Criar e concatenar onda para cada nota
    onda = np.array([])
    f_notas = {'c': 261.63, 'c#': 277.18, 'd': 293.66, 'd#': 311.13,
               'e': 329.63, 'f': 349.23, 'f#': 369.99, 'g': 392.00,
               'g#': 415.30, 'a': 440.00, 'a#': 466.16, 'b': 493.88,
               'c5': 523.25}
    for nota, duracao in notas:
        f = f_notas[nota.lower()]
        duracao_total_nota = duracao_unidade * duracao
        t = np.linspace(0, duracao_total_nota, int(duracao_total_nota * fs), endpoint=False)

        if forma_onda == 'triangular':
            onda_nota = amplitude * scipy.signal.sawtooth(2 * np.pi * f * t, 0.5)
        elif forma_onda == 'dente de serra':
            onda_nota = amplitude * scipy.signal.sawtooth(2 * np.pi * f * t)
        elif forma_onda == 'quadrada':
            onda_nota = amplitude * scipy.signal.square(2 * np.pi * f * t)
        else:
            onda_nota = amplitude * np.sin(2 * np.pi * f * t)
        onda = np.concatenate((onda, onda_nota))
    return onda

output = widgets.Output()

dropdown = widgets.Dropdown(
    options=['Choose a graphic','Onda Sinusoide', 'Onda Triangular', 'Onda Dente de Serra', 'Onda Quadrada'],
    value='Choose a graphic',
    description='Graphics:'
)

def on_dropdown_change(change):
    with output:
        clear_output(wait=True)
        if change['new'] == 'Onda Sinusoide':
             # gerar onda sinusoide (música normal)
            normal = criar_composicao(notas, bpm,'sinusoide')

            # plot onda sinusoide
            plt.figure(figsize=(8, 6))
            plt.plot(normal)
            plt.title('Sinusoide')
            plt.xlim(0,1000)
            plt.xlabel('Tempo (s)')
            plt.ylabel('Música(t)')
            plt.show()

        elif change['new'] == 'Onda Triangular':
            # gerar onda triangular
            onda_triangular = criar_composicao(notas, bpm,'triangular')
            plt.figure(figsize=(8, 6))
            plt.plot(onda_triangular)
            plt.title('Onda Triangular')
            plt.xlim(0,1000)
            plt.xlabel('Tempo (s)')
            plt.ylabel('Música(t)')
            plt.show()


        elif change['new'] == 'Onda Dente de Serra':
            sawtooth = criar_composicao(notas, bpm,'dente de serra')

            # plot onda dente de serra
            plt.figure(figsize=(8, 6))
            plt.plot(sawtooth)
            plt.title('Onda Dente de Serra')
            plt.xlim(0,1000)
            plt.xlabel('Tempo (s)')
            plt.ylabel('Música(t)')
            plt.show()

        elif change['new'] == 'Onda Quadrada':
            # gerar onda quadrada
            square = criar_composicao(notas, bpm,'quadrada')

            # plot onda quadrada
            plt.figure(figsize=(8, 6))
            plt.plot(square)
            plt.title('Onda Quadrada')
            plt.xlim(0,1000)
            plt.xlabel('Tempo (s)')
            plt.ylabel('Música(t)')
            plt.show()
        

dropdown.observe(on_dropdown_change, names='value')

display(dropdown)
display(output)
```

### Nº de km a subir, a descer e em plano dos ciclistas:
#### Como todos os ciclistas fazem o mesmo caminho só se calcula para um participante

```python
dist_subida = 0
dist_descida = 0
dist_em_plano = 0

for i in range(1, len(marco_data['distance'])):
    diferenca = marco_data['distance'][i] - marco_data['distance'][i-1]
    if marco_data['grade_smooth'][i] > 0:
        dist_subida += diferenca
    elif marco_data['grade_smooth'][i] < 0:
        dist_descida += abs(diferenca)
    else:
        dist_em_plano += abs(diferenca)

print('Diatancia a subir:', round(dist_subida * 0.001, 2), 'km')
print('Diatancia em plano:', round(dist_em_plano * 0.001, 2), 'km')
print('Diatancia a descer:', round(dist_descida * 0.001, 2), 'km')
```
